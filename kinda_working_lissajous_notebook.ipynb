```python
import numpy as np
from numpy import pi as pi
import matplotlib.pyplot as plt
```

```python
# define parametric values of angle and phase
# omega_a = 4  # frequency (Hertz) same for input and output 
# omega_b = 4
# phase = 0 # phase angle shift of signal_1 (radians) 
# phi_2 = 0  # phase shift of signal_2 (radians)
```

```python
#define the domain 
resolution = 1  # increase this value for a larger domain for the function to plot
t = np.linspace(0, 2*pi, num  = 360 * resolution)  # domain of the function (x)
print(t)
```

```python
# define the functional relationship  between the 
# signal_1 = A * np.cos( t * omega_a)
# signal_2 = B * np.cos( t * omega_b + phase)
# print(signal_1)
for omega_1 in np.arange(0, 10, 1):
    for omega_2 in np.arange(0, 10, 1):
        # subplot 3: lissajous
        sig_1 = np.cos(omega_1*t)
        sig_2 = np.cos(omega_2*t)
        plt.plot(sig_1, sig_2)
        plt.xlabel('signal_1')
        plt.ylabel('signal_2')
        plt.pause(0.05)
    
    plt.show()
plt.show()
```

```python
## Use the Matplotlib Animation API to update the plot
# n_iter = 180 

for omega_1 in np.arange(0, 10, 1):
    for omega_2 in np.arange(0, 10, 1):
        for phi in np.arange(0, 2*pi):
            # define plot domains
            # t = np.linspace(0.0, 2*pi, pi/360)
            # define signals 1 and 2 to compare
            sig_1 = np.cos(omega_1*t)
            sig_2 = np.cos(omega_2*t + phi)

            plt.title('Four Visualizations of Two Sinusoids of Variable Phase Difference')
            # subplot 1:  time-series plots
            plt.subplot(2,2,1)
            plt.plot(t, sig_1) 
            plt.xlabel('Time (s)')
            plt.ylabel('Signal 1 Voltage')

            # subplot 2: 
            plt.subplot(2,2,2)
            plt.plot(t, sig_2)
            plt.xlabel('time(s)')
            plt.ylabel('Signal 2 Voltage')
            
            # subplot 3: lissajous
            plt.subplot(2,2,3)
            plt.plot(sig_1, sig_2)
            plt.xlabel('signal_1')
            plt.ylabel('signal_2')

            # subplot 4: Phasor
            plt.subplot(2,2,4)
            phase_1 = 0
            phase_2 = phi
            # pt_1, pt_2 = phasortransform(sig_1, sig_2)
            plt.polar(phase_1, phase_2)     
            plt.pause(0.1)
            
        plt.show()
```
