### Python Code to Build a Dictionary of My Cats and Display Their Colours and Breeds

````python
cat_dict = {'Fuzz' : 'Orange Maine Coon', 'Kohaku' : 'Orange Maine Coon',\
            'Kitten' : 'Brown Tabby'}

cat_names = list(cat_dict.keys())
print("My cats' names are %s\n" % cat_names)

cat_types = list(cat_dict.values())
print("Their breeds and colours are %s\n" % cat_types)
````

***
Python Dictionaries (`dicts`) are super-cool features of the language that are really powerful for managing data. From the [Python 3 documentation](https://docs.python.org/3/tutorial/datastructures.html#dictionaries): 

> Another useful data type built into Python is the dictionary (see Mapping Types — dict). Dictionaries are sometimes found in other languages as “associative memories” or “associative arrays”. Unlike sequences, which are indexed by a range of numbers, dictionaries are indexed by keys, which can be any immutable type; strings and numbers can always be keys. Tuples can be used as keys if they contain only strings, numbers, or tuples; if a tuple contains any mutable object either directly or indirectly, it cannot be used as a key. You can’t use lists as keys, since lists can be modified in place using index assignments, slice assignments, or methods like append() and extend().
>
> It is best to think of a dictionary as a set of key: value pairs, with the requirement that the keys are unique (within one dictionary). A pair of braces creates an empty dictionary: {}. Placing a comma-separated list of key:value pairs within the braces adds initial key:value pairs to the dictionary; this is also the way dictionaries are written on output.

To generate the dict, you simply use the following syntax in Python:

```python
example_dict = {'key1': '<value>', 'key2':,'<value>'}

## the dict entries may be accessed with the keys and values methods
keys_list = example_dict.keys()
values_list = example_dict.values()
```
Another neat feature of Python `dict`s is that you can use various datatypes for keys and values. The following table lays out some, but not all, of the options available to you to use for keys and values.

| Object    | `char` | `int` | `string` | `tuple` | `list` | `set` |
| :-------: | :----: | :---: | :------: | :-----: | :----: | :---: |
| **Key**   | yes    | yes   | yes      | yes     |  no    |  no   | 
| **Value** | yes    | yes   | yes      | yes     | yes    | yes   | 

As you might notice, as long as the datatype is **immutable** you can use it for a key entry. Key entries *cannot* be duplicated, however -- think of it like the `PRIMARY KEY` in a database table.
